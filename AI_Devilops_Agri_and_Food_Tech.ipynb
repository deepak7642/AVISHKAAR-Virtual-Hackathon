### ***Ajmer APMC Mandi Forecasting***

```python
df_cotton_ajmer_forecast = df_cotton_ajmer[
    (df_cotton_ajmer['Crop Type'] == 'Cotton') &
    (df_cotton_ajmer['District'] == 'Ajmer') &
    (df_cotton_ajmer['Market'] == 'Ajmer APMC Mandi')
].copy()
```

```python
df_cotton_ajmer_forecast.head(14)
```

```python
#df_cotton_ajmer_forecast['ds'] = pd.to_datetime(df_cotton_ajmer_forecast['Date'], format="%d-%m-%Y", dayfirst=True )
df_cotton_ajmer_forecast['ds'] = pd.to_datetime(df_cotton_ajmer_forecast['Date'])
df_cotton_ajmer_forecast['y'] = df_cotton_ajmer_forecast['Price (INR/quintal)']
```

```python
#regressors = [col for col in regressors if col in df_cotton_ajmer_forecast.columns]
```

```python
from prophet import Prophet

# Make sure regressors list is valid
regressors = ['Soil pH','Soil Quality','Water Consumption (liters/hectare)','Water Availability (liters/hectare)', 'Temperature', 'Humidity','Wind_Speed', 'N', 'P', 'K', 'Crop Yield']

m = Prophet()

# Add all numeric regressors
for reg in regressors:
    m.add_regressor(reg)

# Fit model (must include ALL regressor columns)
m.fit(df_cotton_ajmer_forecast[['ds', 'y'] + regressors])
```

```python
future = m.make_future_dataframe(periods=15)   #  days forecast

# Add regressors to the future frame
for reg in regressors:
    future[reg] = df_cotton_ajmer_forecast[reg].iloc[-1]      # using last value


forecast = m.predict(future)
```

```python
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']]
```

```python
forecast
```

```python
from sklearn.metrics import mean_absolute_percentage_error

y_true = df_cotton_ajmer_forecast['Price (INR/quintal)'][-730:]
y_pred = forecast['yhat'][:730]

mape = mean_absolute_percentage_error(y_true, y_pred)
print(f"MAPE for Cotton Prices for Ajmer APMC Mandi in Ajmer district : {mape * 100:.2f}%")
```

```python
import pandas as pd


df = pd.read_csv('/content/Merge agri data.csv')
df
```

```python
import pandas as pd
from ipywidgets import interact_manual, Dropdown, Text
from IPython.display import display

# Ensure Date is datetime
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

def display_7day_forecast(crop, district, start_date_str):
    try:
        start_date = pd.to_datetime(start_date_str)
    except:
        print("❌ Invalid date format — use YYYY-MM-DD")
        return

    end_date = start_date + pd.Timedelta(days=7)

    # Filter dataframe
    filtered_df = df[
        (df['Crop Type'] == crop) &
        (df['District'] == district) &
        (df['Date'] >= start_date) &
        (df['Date'] <= end_date)
    ]

    if filtered_df.empty:
        print("⚠ No forecast data available for this selection / date range.")
        return

    # Drop duplicates for Date + Market to avoid repeated rows
    filtered_df = filtered_df.drop_duplicates(subset=['Date', 'Market'])

    # Sort by Date and Market
    filtered_df = filtered_df.sort_values(['Date', 'Market']).reset_index(drop=True)

    # Display only relevant columns
    display(filtered_df[['Date', 'Market', 'Forecasted_Price']])

# Dropdowns
crop_dropdown = Dropdown(
    options=sorted(df['Crop Type'].unique()),
    description='Crop:'
)

district_dropdown = Dropdown(
    options=sorted(df['District'].unique()),
    description='District:'
)

date_input = Text(
    value='2014-01-01',
    description='Start Date (YYYY-MM-DD):'
)

# Run button
interact_manual(display_7day_forecast, crop=crop_dropdown, district=district_dropdown, start_date_str=date_input)
```

## **Gradio Deployment**

```python
import pandas as pd
import gradio as gr

# Ensure Date column is datetime
#df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

def display_7day_forecast_gradio(crop, district, start_date_str):
    try:
        start_date = pd.to_datetime(start_date_str)
    except:
        return "❌ Invalid date format — use YYYY-MM-DD"

    end_date = start_date + pd.Timedelta(days=7)

    # Filter dataframe
    filtered_df = df[
        (df['Crop Type'] == crop) &
        (df['District'] == district) &
        (df['Date'] >= start_date) &
        (df['Date'] <= end_date)
    ]

    if filtered_df.empty:
        return "⚠ No forecast data available for this selection / date range."

    # Drop duplicates and sort
    filtered_df = filtered_df.drop_duplicates(subset=['Date', 'Market'])
    filtered_df = filtered_df.sort_values(['Date', 'Market']).reset_index(drop=True)

    # Return relevant columns
    return filtered_df[['Date', 'Market', 'Forecasted_Price']]

# Gradio Interface
crop_list = sorted(df['Crop Type'].unique())
district_list = sorted(df['District'].unique())

iface = gr.Interface(
    fn=display_7day_forecast_gradio,
    inputs=[
        gr.Dropdown(choices=crop_list, label="Crop"),
        gr.Dropdown(choices=district_list, label="District"),
        gr.Textbox(value="2014-01-01", label="Start Date (YYYY-MM-DD)")
    ],
    outputs="dataframe",
    title="7-Day Forecast Table",
    description="Select Crop and District, enter a start date, and see the next 7 days forecasted prices."
)

iface.launch()
```
